### Python API

- `VLLMOpenAI`: 제어문장 (`<|im_start|>`, `<|im_end|>`, ...) 자동화 안 됨

```python
from langchain_community.llms.vllm import VLLMOpenAI, VLLM
from langchain.prompts import ChatPromptTemplate


llm_server_url = "http://localhost:58000/v1"
llm = VLLMOpenAI(
    openai_api_key="EMPTY",
    openai_api_base=llm_server_url,
    model_name="Qwen/Qwen3-14B-AWQ",
    temperature=0.7,
    top_p=0.8,
    max_tokens=50,  # 8192
    presence_penalty=1.5,
    extra_body={"top_k": 20, "chat_template_kwargs": {"enable_thinking": False, "quantization": "awq"}},
)

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("user", "{input}")
])

# Thinking: False
input = prompt_template.invoke({"input": "넌 누구니?"})
print(llm.invoke(input, extra_body={"chat_template_kwargs": {"enable_thinking": False}}))

print(100*"=")

# Thinking: True
input = prompt_template.invoke({"input": "좀 더 자세히 말해봐"})
print(llm.invoke(input, extra_body={"chat_template_kwargs": {"enable_thinking": True}}))
```

- `ChatOpenAI`: 제어문장 (`<|im_start|>`, `<|im_end|>`, ...) 자동화됨

```python
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate


llm_server_url = "http://localhost:58000/v1"
llm = ChatOpenAI(
    openai_api_key="EMPTY",
    openai_api_base=llm_server_url,
    model_name="Qwen/Qwen3-14B-AWQ",
    temperature=0.7,
    top_p=0.8,
    max_tokens=10,
    presence_penalty=1.5,
    # Default thinking: False
    extra_body={"top_k": 20, "chat_template_kwargs": {"enable_thinking": False, "quantization": "awq"}},
    seed=0,
)

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("user", "{input}")
])

# Thinking: False
input = prompt_template.invoke({"input": "넌 누구니?"})
print(llm.invoke(input, extra_body={"chat_template_kwargs": {"enable_thinking": False}}).content)

print(100*"=")

# Thinking: True
input = prompt_template.invoke({"input": "좀 더 자세히 말해봐"})
print(llm.invoke(input, extra_body={"chat_template_kwargs": {"enable_thinking": True}}).content)
```
